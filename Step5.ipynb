```python
# 读取HauntedPlaces CSV数据集
merged_Crim = pd.read_csv('merged_df.csv')
```

```python
# 读取地震数据集 (JSON)
with open('earthquake_data(1).json', 'r') as f:
    eq_data = json.load(f)

# 转换地震数据为DataFrame，提取有用信息
eq_list = []
for feature in eq_data['features']:
    eq_coords = feature['geometry']['coordinates']  # 经度,纬度,深度
    eq_properties = feature['properties']
    eq_list.append({
        'longitude': eq_coords[0],
        'latitude': eq_coords[1],
        'mag': eq_properties['mag'],
        'time': datetime.fromtimestamp(eq_properties['time']/1000)  # 毫秒转换
    })

eq_df = pd.DataFrame(eq_list)
eq_df
```

```python
def calculate_earthquake_features(row, eq_df, radius_km=20):
    place_coords = (row['latitude'], row['longitude'])
    
    # 创建一个副本来计算距离，避免修改原始数据
    eq_df_copy = eq_df.copy()

    # 计算距离
    eq_df_copy['distance'] = eq_df_copy.apply(
        lambda x: geodesic(place_coords, (x['latitude'], x['longitude'])).km, axis=1
    )

    # 筛选指定范围内的地震
    nearby_eq = eq_df_copy[eq_df_copy['distance'] <= radius_km]

    # 提取特征
    earthquake_count = len(nearby_eq)
    max_mag = nearby_eq['mag'].max() if earthquake_count > 0 else 0
    latest_time = nearby_eq['time'].max() if earthquake_count > 0 else pd.NaT
    
    return pd.Series({
        'recent_earthquake_count': earthquake_count,
        'recent_earthquake_max_mag': max_mag,
        'recent_earthquake_latest_time': latest_time
    })

```

```python
# 先过滤掉HauntedPlaces中经纬度缺失的数据
merged_Crim = merged_Crim.dropna(subset=['latitude', 'longitude'])
```

```python

# 高效计算地震特征
def calculate_optimized_features(row, eq_df, radius_km=20):
    place_coords = (row['latitude'], row['longitude'])

    # 使用向量化预过滤
    lat_min, lat_max = row['latitude'] - 0.2, row['latitude'] + 0.2
    lon_min, lon_max = row['longitude'] - 0.2, row['longitude'] + 0.2

    nearby_eq_df = eq_df[
        (eq_df['latitude'] >= lat_min) & (eq_df['latitude'] <= lat_max) &
        (eq_df['longitude'] >= lon_min) & (eq_df['longitude'] <= lon_max)
    ].copy()

    # 如果没有符合的，直接返回空特征
    if nearby_eq_df.empty:
        return pd.Series({
            'recent_earthquake_count': 0,
            'recent_earthquake_max_mag': None,
            'recent_earthquake_latest_time': None
        })

    # 精确计算距离并进一步筛选20km内的地震
    nearby_eq_df['distance'] = nearby_eq_df.apply(
        lambda x: geodesic(place_coords, (x['latitude'], x['longitude'])).km, axis=1
    )
    nearby_eq_df = nearby_eq_df[nearby_eq_df['distance'] <= 20]

    # 提取特征
    earthquake_count = len(nearby_eq_df)
    recent_earthquake_max_mag = nearby_eq_df['mag'].max()
    recent_earthquake_latest_time = nearby_eq_df['properties.time'].max()
    recent_earthquake_latest_time = pd.to_datetime(recent_earthquake_latest_time, unit='ms')

    return pd.Series({
        'recent_earthquake_count': earthquake_count,
        'recent_earthquake_max_mag': recent_earthquake_max_mag,
        'recent_earthquake_latest_time': recent_earthquake_latest_time
    })

# 应用到数据集
earthquake_features = merged_Crim.apply(
    lambda row: calculate_earthquake_features(row, eq_df), axis=1
)

# 合并特征到HauntedPlaces数据集中
merged_Crim = pd.concat([merged_Crim.reset_index(drop=True), earthquake_features.reset_index(drop=True)], axis=1)
```
